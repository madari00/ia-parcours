```python
import numpy as np
import matplotlib.pyplot as plt

# Données synthétiques : y = 3x + 2 + bruit
rng = np.random.default_rng(0)
X = rng.uniform(0, 10, size=(200, 1))
y = 3 * X[:, 0] + 2 + rng.normal(0, 1, size=200)
print("X:", X.shape, "| y:", y.shape)
```
